```python
data = pd.read_csv("ecommerce_sales_data.csv")

print("Dataset loaded successfully!")
data.head()
```

```python
data["Category"].value_counts().plot(kind="bar")

plt.title("Number of Records by Category")
plt.xlabel("Category")
plt.ylabel("Count")
plt.xticks(rotation=45)
plt.show()
```

```python
category_revenue = data.groupby("Category")["Revenue"].sum()

category_revenue.sort_values().plot(kind="barh")

plt.title("Total Revenue by Category")
plt.xlabel("Revenue")
plt.ylabel("Category")
plt.show()
```

```python
plt.figure(figsize=(10, 6))

sns.barplot(
    data=data,
    x="Region",
    y="Revenue",
    hue="Category",
    errorbar=None
)

plt.title("Revenue by Region and Category")
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=data,
    x="Ad_Spend",
    y="Profit",
    hue="Region"
)

plt.title("Advertising Spend vs Profit")
plt.show()
```

```python
print("Highest Revenue Category:")
print(data.groupby("Category")["Revenue"].sum().idxmax())

print("\nHighest Revenue Region:")
print(data.groupby("Region")["Revenue"].sum().idxmax())

print("\nHighest Profit Category:")
print(data.groupby("Category")["Profit"].sum().idxmax())

print("\nTotal Revenue:")
print(data["Revenue"].sum())

print("\nTotal Profit:")
print(data["Profit"].sum())
```
